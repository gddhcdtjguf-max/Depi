```python
#TASK1
email = "Amit_ml@gmail.edu"

# Validation
if email.count("@") != 1:
    print("Invalid email")
else:
    at = email.index("@")

    if "." not in email[at:]:
        print("Invalid email")
    else:
        # Username
        username = email[:at]

        # Domain
        domain = email[at+1:email.rfind(".")]

        print("Username:", username)
        print("Domain:", domain)

        # Domain type
        if email.endswith(".com"):
            print("Commercial Domain")
        elif email.endswith(".edu"):
            print("Educational Domain")
        else:
            print("Other Domain")
```

```python
#TASK2
message = "###!!@mocleW EPGTQ!!!6789"

core = ""

for ch in message:
    if ch.isalpha() or ch == " ":
        core += ch

words = core.split()

first = words[0][::-1]
second = words[1]

print(first, second)
```

```python
#TASK3
message = "&&&**$gnirtS PLIO!!@1234"

core = ""

for ch in message:
    if ch.isalpha() or ch == " ":
        core += ch

words = core.split()

first = words[0][::-1]

second = ""

for ch in words[1]:
    if ch == "I":
        second += "E"
    elif ch == "O":
        second += "U"
    else:
        second += ch

print(first, second)
```

```python
#TASK4
message = "##$$$@!yalpstcejorp EPUVT****9887"

# Extract letters and spaces only
core = ""

for ch in message:
    if ch.isalpha() or ch == " ":
        core += ch

# Split into words
words = core.split()

# Reverse first word
first = words[0][::-1]

# Replace vowels in second word
second = ""

for ch in words[1]:
    if ch == "E":
        second += "A"
    elif ch == "U":
        second += "O"
    else:
        second += ch

# Print result
print(first, second)
```
