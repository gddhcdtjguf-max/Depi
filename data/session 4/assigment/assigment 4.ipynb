```python
def multiplication_table(num):
    for i in range(1, 11):
        print(f"{num} x {i} = {num * i}")

n = int(input("Enter a number: "))
multiplication_table(n)
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


for i in range(2, 1000):
    if is_prime(i) and is_prime(i + 2):
        print(i, i + 2)
```

```python
def prime_factors(n):
    i = 2

    while i <= n:
        while n % i == 0:
            print(i)
            n = n // i
        i += 1

num = int(input("Enter a number: "))
prime_factors(num)
```

```python
def decimal_to_binary(n):
    binary = ""

    while n > 0:
        binary = str(n % 2) + binary
        n = n // 2

    print(binary)

num = int(input("Enter a decimal number: "))
decimal_to_binary(num)
```

```python
def is_perfect(n):
    total = 0

    for i in range(1, n):
        if n % i == 0:
            total += i

    return total == n


start = int(input("Enter start: "))
end = int(input("Enter end: "))

for i in range(start, end + 1):
    if is_perfect(i):
        print(i)
```

```python
import os
import random
"""
    Creates a number of files inside a folder, then randomly deletes
    half of them and prints the result before and after.

    :param folder_path: The path of the folder to create/use.
    :type folder_path: str
    :param num_files: The number of files to create inside the folder.
    :type num_files: int

    :raise FileNotFoundError: If the folder path is invalid.

    :return: None, the function only prints the results.
    :rtype: None
    """

def thanos_project(folder_path, num_files):

    if not os.path.exists(folder_path):
        os.makedirs(folder_path)

    for i in range(1, num_files + 1):
        file_path = os.path.join(folder_path, f"file{i}.txt")
        with open(file_path, "w") as f:
            f.write(f"This is file {i}")
    files = os.listdir(folder_path)
    print("Files before deleting:", len(files))

    random_files = random.sample(files, len(files) // 2)

    for file in random_files:
        os.remove(os.path.join(folder_path, file))

    files = os.listdir(folder_path)
    print("Files after deleting:", len(files))


folder = "ThanosFolder"
number = int(input("Enter number of files: "))

thanos_project(folder, number)
```

```python
def get_number(prompt):
    """
    Keeps asking the user for a number until a valid one is entered.

    :param prompt: The message shown to the user.
    :type prompt: str

    :return: A valid float number entered by the user.
    :rtype: float
    """
    while True:
        user_input = input(prompt)
        try:
            return float(user_input)
        except ValueError:
            print("الرقم اللي دخلته مش صح، حاول تاني.")


def calculate(operation, num1, num2):
    """
    Performs the selected arithmetic operation on two numbers.

    :param operation: The operation chosen by the user (1 to 4).
    :type operation: str
    :param num1: The first number.
    :type num1: float
    :param num2: The second number.
    :type num2: float

    :raise ZeroDivisionError: If operation is division and num2 is zero.

    :return: The result of the operation, or None if division by zero.
    :rtype: float or None
    """
    if operation == "1":
        return num1 + num2
    elif operation == "2":
        return num1 - num2
    elif operation == "3":
        return num1 * num2
    elif operation == "4":
        if num2 == 0:
            print("مينفعش تقسم على صفر!")
            return None
        return num1 / num2


def basic_calculator():
    """
    Runs the main loop of the calculator program:
    shows the menu, gets user choice, performs calculation,
    and asks if the user wants to continue.

    :return: None
    :rtype: None
    """
    operations_names = {
        "1": "adding",
        "2": "subtracting",
        "3": "multiplying",
        "4": "dividing"
    }

    while True:
        print("\nاختار العملية:")
        print("1- Addition")
        print("2- Subtraction")
        print("3- Multiplication")
        print("4- Division")
        print("5- Exit")

        choice = input("اختيارك: ")

        if choice == "5":
            print("شكراً لاستخدامك الآلة الحاسبة!")
            break

        if choice not in ["1", "2", "3", "4"]:
            print("اختيار مش موجود، حاول تاني.")
            continue

        num1 = get_number("ادخل الرقم الأول: ")
        num2 = get_number("ادخل الرقم الثاني: ")

        result = calculate(choice, num1, num2)

        if result is not None:
            op_name = operations_names[choice]
            print(f"The result of {op_name} {num1} and {num2} is {result}")

        again = input("\nعايز تعمل حساب تاني؟ (yes/no): ").lower()
        if again != "yes":
            print("شكراً لاستخدامك الآلة الحاسبة!")
            break


# تشغيل البرنامج
basic_calculator()
```
