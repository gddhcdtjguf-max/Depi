```python
class MultiplicationTable:

    def __init__(self, num):
        self.num = num

    def multiplication_table(self):
        for i in range(1, 11):
            print(f"{self.num} x {i} = {self.num * i}")


n = int(input("Enter a number: "))

table = MultiplicationTable(n)
table.multiplication_table()
```

```python
class PrimeNumbers:

    def __init__(self, limit):
        self.limit = limit

    def is_prime(self, n):
        if n < 2:
            return False

        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False

        return True

    def find_twin_primes(self):
        for i in range(2, self.limit):
            if self.is_prime(i) and self.is_prime(i + 2):
                print(i, i + 2)


n = int(input("Enter the limit: "))

prime = PrimeNumbers(n)
prime.find_twin_primes()
```

```python
class PrimeFactors:

    def __init__(self, n):
        self.n = n

    def prime_factors(self):
        i = 2

        while i <= self.n:
            while self.n % i == 0:
                print(i)
                self.n = self.n // i

            i += 1


num = int(input("Enter a number: "))

factors = PrimeFactors(num)
factors.prime_factors()
```

```python
class DecimalToBinary:

    def __init__(self, n):
        self.n = n

    def convert(self):
        binary = ""

        while self.n > 0:
            binary = str(self.n % 2) + binary
            self.n = self.n // 2

        print(binary)


num = int(input("Enter a decimal number: "))

binary_num = DecimalToBinary(num)
binary_num.convert()
```

```python
class PerfectNumber:

    def __init__(self, start, end):
        self.start = start
        self.end = end

    def is_perfect(self, n):
        total = 0

        for i in range(1, n):
            if n % i == 0:
                total += i

        return total == n

    def find_perfect_numbers(self):
        for i in range(self.start, self.end + 1):
            if self.is_perfect(i):
                print(i)


start = int(input("Enter start: "))
end = int(input("Enter end: "))

numbers = PerfectNumber(start, end)
numbers.find_perfect_numbers()
```

```python
import os
import random


class ThanosProject:

    def __init__(self, folder_path, num_files):
        self.folder_path = folder_path
        self.num_files = num_files

    def create_files(self):
        if not os.path.exists(self.folder_path):
            os.makedirs(self.folder_path)

        for i in range(1, self.num_files + 1):
            file_path = os.path.join(
                self.folder_path, f"file{i}.txt"
            )

            with open(file_path, "w") as f:
                f.write(f"This is file {i}")

    def delete_files(self):
        files = os.listdir(self.folder_path)

        print("Files before deleting:", len(files))

        random_files = random.sample(files, len(files) // 2)

        for file in random_files:
            os.remove(os.path.join(self.folder_path, file))

        files = os.listdir(self.folder_path)

        print("Files after deleting:", len(files))


folder = "ThanosFolder"
number = int(input("Enter number of files: "))

thanos = ThanosProject(folder, number)

thanos.create_files()
thanos.delete_files()
```

```python
class BasicCalculator:

    def __init__(self):
        self.operations_names = {
            "1": "adding",
            "2": "subtracting",
            "3": "multiplying",
            "4": "dividing"
        }

    def get_number(self, prompt):
        while True:
            user_input = input(prompt)

            try:
                return float(user_input)

            except ValueError:
                print("الرقم اللي دخلته مش صح، حاول تاني.")

    def calculate(self, operation, num1, num2):

        if operation == "1":
            return num1 + num2

        elif operation == "2":
            return num1 - num2

        elif operation == "3":
            return num1 * num2

        elif operation == "4":

            if num2 == 0:
                print("مينفعش تقسم على صفر!")
                return None

            return num1 / num2

    def basic_calculator(self):

        while True:

            print("\nاختار العملية:")
            print("1- Addition")
            print("2- Subtraction")
            print("3- Multiplication")
            print("4- Division")
            print("5- Exit")

            choice = input("اختيارك: ")

            if choice == "5":
                print("شكراً لاستخدامك الآلة الحاسبة!")
                break

            if choice not in ["1", "2", "3", "4"]:
                print("اختيار مش موجود، حاول تاني.")
                continue

            num1 = self.get_number("ادخل الرقم الأول: ")
            num2 = self.get_number("ادخل الرقم الثاني: ")

            result = self.calculate(choice, num1, num2)

            if result is not None:
                op_name = self.operations_names[choice]

                print(
                    f"The result of {op_name} "
                    f"{num1} and {num2} is {result}"
                )

            again = input(
                "\nعايز تعمل حساب تاني؟ (yes/no): "
            ).lower()

            if again != "yes":
                print("شكراً لاستخدامك الآلة الحاسبة!")
                break


# Create object
calculator = BasicCalculator()

# Run calculator
calculator.basic_calculator()
```
