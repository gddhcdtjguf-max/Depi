```python
class employee:
    def __init__(self,name,salary):
        self.__salary = salary
        self.name = name
    def set_salary(self,obj, new_salary):
        obj.__salary = new_salary

      
    class manager:
        def __init__(self,name,salary):
            self.name = name
            self__salary = salary
         
```

```python
e1 = employee("depi", 1000)
m1 = employee.manager("amit", 2000)
e1.set_salary(e1, 2000)
print(m1.salary)
```

```python
class A():
    def do_this(self):
        print("do this from A")
class B(A):
     def do_this(self):
         print("do this from B")
class C():
     def do_this(self):
         print("do this from C")
class D(B,C):
         pass
x=D()
x.do_this()
```
